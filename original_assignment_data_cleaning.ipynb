```python
# Create a copy of the original dataset
clean_data = df.copy()

# Remove unnecessary index column
clean_data = clean_data.drop("Unnamed: 0", axis=1)

# Remove extra spaces from column names
clean_data.columns = clean_data.columns.str.strip()

# Remove extra spaces from text columns
for column in clean_data.select_dtypes(include="object").columns:
    clean_data[column] = clean_data[column].str.strip()

print(clean_data.head())
```

```python
print(clean_data["Best Time to visit"].value_counts())
```

```python
print(clean_data["Weekly Off"].value_counts(dropna=False))
```

```python
print(clean_data[clean_data["Weekly Off"] == "Yes"])
```

```python
# Replace incorrect Weekly Off value
clean_data.loc[clean_data["Weekly Off"] == "Yes", "Weekly Off"] = "Not Specified"

# Replace missing Weekly Off values
clean_data["Weekly Off"] = clean_data["Weekly Off"].fillna("Not Specified")

print(clean_data["Weekly Off"].value_counts())
```

```python
numeric_columns = [
    "Google review rating",
    "Entrance Fee in INR",
    "time needed to visit in hrs",
    "Number of google review in lakhs"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )
```

```python
clean_data.to_csv(
    "cleaned_tourist_data.csv",
    index=False
)
```
